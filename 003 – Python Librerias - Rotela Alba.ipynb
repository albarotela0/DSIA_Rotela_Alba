## Ejercicio 2 – Dataset simple con Pandas

```
#Estructura de datos
datos_estudiantes = {
    "estudiante": ["Ana", "Juan", "Carla", "Pedro", "Lucia"],
    "nota": [8, 6, 9, 5, 7]
}

#DataFrame de Pandas
df = pd.DataFrame(datos_estudiantes)

#Mostrar el DataFrame completo
print("DataFrame de estudiantes")
print(df)

#Cálculos
promedio = df["nota"].mean()
nota_maxima = df["nota"].max()
nota_minima = df["nota"].min()

#Mostrar los resultados(aunque el promedio no se pedía)
print("Promedio de notas:", promedio)
print("Nota máxima:", nota_maxima)
print("Nota mínima:", nota_minima)
```

## Ejercicio 3 – Criterio de selección de herramientas

¿Qué librería utilizarías para cada una de las siguientes tareas?

1\. Leer un archivo CSV con datos de encuestas
2\. Crear un gráfico de distribución de datos
3\. Realizar un modelo de Machine Learning
4\. Trabajar con operaciones matemáticas sobre matrices

pandas → porque es la librería principal para cargar archivos CSV y organizar los datos en tablas \(DataFrames\) fáciles de manipular\.

seaborn → porque está especializada en gráficos estadísticos y permite mostrar la distribución de los datos de forma rápida y visual\.
