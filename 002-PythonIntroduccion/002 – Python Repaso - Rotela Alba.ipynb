## Ejercicio 1 \- Variables y operaciones

```
#Parte 1
nombre = "Alba"
edad = 33
ciudad = "Narnia"

print(f"Hola, mi nombre es {nombre}, tengo {edad} años y vivo en {ciudad}.")

#Parte 2
a = 20
b = 10

suma = a + b
resta = a - b
multiplicacion = a * b
division = a / b

print(f"Suma: {a} + {b} = {suma}")
print(f"Resta: {a} - {b} = {resta}")
print(f"Multiplicación: {a} * {b} = {multiplicacion}")
print(f"División: {a} / {b} = {division}")
```

## Ejercicio 2 \- Condicionales

```
#Pida al usuario que ingrese su edad usando input()
edad_usuario = int(input("Ingresa tu edad: "))

#Determine si la persona es mayor o menor de edad
if edad_usuario >= 18:
    print("Eres mayor de edad")
else:
    print("Eres menor de edad")
```

## Ejercicio 3 \- Listas y bucles

```
#Crear una lista con 5 números.
numeros = [1, 2, 3, 4, 5]

#Utilizando un bucle for, imprimir cada número de la lista.
for num in numeros:
    print(num)

#Calcular la suma total de los números
suma_total = sum(numeros)

#Calcular el promedio
promedio = suma_total / len(numeros)

#Mostrar ambos resultados por pantalla.
print(f"Suma total: {suma_total}")
print(f"Promedio: {promedio}")
```
